space must be sorted for binary search- keeps dividing in half and only searching a smaller piece at a time  
log n = cut search space in half until we get the the last one element

```python
# recursive linear search algorithm
def linear_search(nlist: int, target):
  if len(nlist) == 0:
    return f"{target} not found in list."

  if target == nlist[0]:
    return f"{target} found in list"
  else:
    nlist.pop(0)
    return linear_search(nlist, target)

print(linear_search([1, 2, 3, 4, 5], 3))
print(linear_search([1, 2, 3, 4, 5], 5))
print(linear_search([1, 2, 3, 4, 5], 70))
```

```python
def binary_search(sorted_nums:list, target:int, low:int, high:int):
  #base case
  if low > high:
    return -1

  mid = (low + high) // 2  # integer division = floor

  # if found
  if sorted_nums[mid] == target:
    print("target found")
    return mid

  elif target < sorted_nums[mid]:
    return binary_search(sorted_nums, target, low, mid-1)  # should not look at midpoint once we checked it already

  else:
    return binary_search(sorted_nums, target, mid+1, high)

print(binary_search([20, 30, 40, 50, 60], 50, 0, 4))
```

merge sort- divide and conquer algorithm  
hint: a list of one element is always sorted = base case  
  - wind until hit base case and then algorithm realizes that it is sorted = divide
  - unwind (productive) starts by combining two elements at a time sorting them and then merge each group with each other until full list is back together = merge

```python
def merge(left_list:list, right_list:list):
  merged = []
  right_index = 0
  left_index = 0

  while left_index < len(left_list) and right_index < len(right_list):
    if left_list[left_index] < right_list[right_index]:
      merged.append(left_list[left_index])
      left_index += 1
    else:
      merged.append(right_list[right_index])
      right_index += 1

  # add remaining values in one list when other list is empty
  while left_index < len(left_list):
    merged.append(left_list[left_index])
    left_index += 1

  while right_index < len(right_list):
    merged.append(right_list[right_index])
    right_index += 1

  return merged

def merge_sort(nums:list):
  if len(nums) <= 1:
    return nums

  mid = len(nums) // 2  # floor

  # wind
  left_list = merge_sort(nums[:mid])
  right_list = merge_sort(nums[mid:])

  # return result so previous stack frame has the values
  return merge(left_list, right_list)


print(merge_sort([50, 30, 20, 60, 70]))
```
