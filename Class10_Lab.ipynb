```python
class Person:
  number_of_people = 100  # class attributes never get overwritten, can see parent version from here if requsted and child attribute from child class
  def __init__(self, name):
    self.name = name
    Person.number_of_people += 1

  def speak(self):
    return "Parent says Hello"

  @classmethod
  def printing_num_people(cls):
    print(cls.number_of_people)

class Student(Person):
  number_of_people = 500
  def __init__(self, name):
    super().__init__(name)
    print(Student.number_of_people)

  def study(self):
    return "Student is studying very hard"

  def speak(self):
    return f"Student says Hello AND  {super().number_of_people}"

  @classmethod
  def how_many_people(cls):
    print(f"{Student.number_of_people} ---- {super().number_of_people}")

s1 = Student("Yocheved")
print(Person.printing_num_people())
print(s1.how_many_people())
print(s1.speak())
```
