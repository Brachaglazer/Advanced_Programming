```python
#non productive unwind- action happens before function call
#no behavior happening in unwind
def message(count:int=5):
  if count > 0:
    print("This is the current count", count)
    count -= 1
    message(count)

  if __name__ == "__main__":
    message(5)
```

```python
#productive unwind- behavior happens in unwind
def countdown(count):
  if count == 0:
    return
  countdown(count - 1)
  print(count)
```

```python
#factorial

def factorial(n:int=5):
  #base case
  if n == 0:
    return 1

  #recursive case
  return n * factorial(n-1) #n = number in current stack frame | return = return to previous stack frame

print(factorial(5))
```

```python
#count how many even numbers we have in a list
def count_even(nlist, index):
  if index == 0:
    if nlist[index] % 2 == 0:
      return 1
    else:
      return 0

  if nlist[index] % 2 == 0:
    return 1 + count_even(nlist, index - 1)
  else:
    return 0 + count_even(nlist, index - 1)

count_even(nlist, len(nlist) - 1)
```
