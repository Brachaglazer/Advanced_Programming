```python
class Pet:
  country = "USA"
  def __init__(self, name: str, age: int, owner: str) -> None:
    self.name = name
    self.age = age
    self.owner = owner

  #getters here
  def get_name(self):
    return self.name
  def get_age(self):
    return self.age
  def get_owner(self):
    return self.owner

  #setters here
  def set_name(self, new_name):
    self.name = new_name
  def set_age(self, new_age):
    self.age = new_age
  def set_owner(self, new_owner):
    self.owner = new_owner

pet1 = Pet("Snowy", 5, "Alex")
pet2 = Pet("Spotty", 7, "Kim")
print(pet1.get_name())
pet2.set_name("Lilly")
print(pet2.get_name())
Pet.country = "EU"
print(pet1.country)
print(pet2.country)

pet1.country ="Italy"
print(pet1.country)
print(pet2.country)
```

```python
class Vet:
  def __init__(self, name: str, store: str, client: Pet):
    self.__name = name
    self.__store = store
    self.__client_name = client.get_owner()
    self.__pet_name = client.get_name()

  def get_name(self):
    return self.__name
  def get_store(self):
    return self.__store
  def get_client_name(self):
    return self.__client_name
  def get_pet_name(self):
    return self.__pet_name

  def set_name(self, new_name):
    self.__name = new_name
  def set_store(self, new_store):
    self.__store = new_store
  def set_client_name(self, new_client):
    self.__client_name = new_client
  def set_pet_name(self, new_pet):
    self.__pet_name = new_pet

  def __str__(self):
    return f"Client's name is {self.__client_name} \nPet's name is {self.__pet_name}"


dr_rayne = Vet("Dr. Rayne", "Pet Shop", pet1)
print(dr_rayne.get_client_name(), dr_rayne.get_pet_name())
print(dr_rayne)
```
