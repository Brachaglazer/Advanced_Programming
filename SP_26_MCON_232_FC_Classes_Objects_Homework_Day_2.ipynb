## Big Reminders
A **class** is a blueprint for making objects.  
An **object** is one instance of that class.  
The constructor `__init__` gives each object its starting values.  
Instance attributes usually belong to one object.  
Class attributes belong to the class and are shared unless shadowed.  
Methods let objects do work.  
One object can receive another object as a parameter and use that object's data.  
If a method mutates an object, that mutation is still there when the method returns.  
If a method reassigns the parameter variable to a different object, the original caller's object does not get replaced.

## Task 8 — Reassignment inside a method

Create two classes: `Plant` and `Gardener`.

### `Plant`
- constructor parameters:
  - `name`
  - `height`

### `Gardener`
Write two methods:

1. `water(plant_obj, inches)`
   - increase `plant_obj.height` by `inches`

2. `replace_local(plant_obj)`
   - reassign `plant_obj` to a new `Plant("Replacement", 100)`

### After writing the classes
- create one original plant object
- call `water(...)`
- print the original plant's data
- call `replace_local(...)`
- print the original plant's data again

### Reflection
In a comment, explain:
- why `water(...)` changes the original object
- why `replace_local(...)` does not replace the original object

````python

# Write your answer here
class Plant:
  def __init__(self, name: str, height: int) -> None:
    """This method initializes the attributes of the class"""
    self.name = name
    self.height = height

  def __str__(self):
    """This method prints information about the object of the class"""
    return f"{self.name} health: {self.height}"

class Gardener():
  def water(self, plant_obj: Plant, inches: int):
    """This method increases the height of the plant object"""
    plant_obj.height += inches

  def replace_local(self, plant_obj: Plant):
    """This method creates a temporary object for our variable to point to"""
    plant_obj = Plant("Replacement", 100)

original_plant = Plant("Lily", 10)
gardener = Gardener()
gardener.water(original_plant, 15)
print(original_plant)
gardener.replace_local(original_plant)
print(original_plant)

# We are passing objects by reference here so when the object is reassigned it simply points to a new variable and does not make any changes to the original variable passed in.
# When mutating a refernce to an object passed in, the original value is changes and remains.
````

## Task 9 — A `BankAccount` class

Create a class named `BankAccount`.

### Requirements
- constructor parameters:
  - `owner`
  - `balance`
- method `deposit(amount)`
  - add to the balance
- method `withdraw(amount)`
  - if enough money exists, subtract and return `True`
  - otherwise do not change the balance and return `False`
- method `display()`
  - return a string like:
    - `"Ari has $120."`

Create one account and test all methods.

### Reminder
This is classic object-oriented design: the object stores state and methods manage that state.

````python

# Write your answer here
class BankAccount:
  def __init__(self, owner: str, balance: int) -> None:
    """This method initializes the attributes of the class"""
    self.owner = owner
    self.balance = balance

  def deposit(self, amount: int):
    """This method decreases the balance if the balance is greater than or equal to the amount"""
    if amount <= self.balance:
      self.balance -= amount
      return True
    return False

  def display(self):
    """This method returns the information about a specific object"""
    return f"{self.owner} has ${self.balance}"

chase = BankAccount("Larry", 100)
chase.deposit(53)
print(chase.display())
````

## Task 10 — A `Classroom` and `Student`

Create two classes: `Student` and `Classroom`.

### `Student`
- constructor parameters:
  - `name`
  - `grade`

### `Classroom`
- constructor should create an empty list named `students`
- method `add_student(student_obj)`
  - add the object to the list
- method `highest_grade()`
  - return the highest grade among the students
  - if there are no students, return `None`

### Test
- create at least 3 students
- add them to a classroom
- print the highest grade

### Reminder
This is good practice with objects containing collections of other objects.

````python

# Write your answer here
class Student:
  def __init__(self, name: str, grade: int) -> None:
    """This method initializes the attributes of the class"""
    self.name = name
    self.grade = grade

class Classroom:
  def __init__(self):
    self.students = []

  def add_student(self, student_obj: Student):
    """This method adds a student object to the student list"""
    self.students.append(student_obj)

  def highest_grade(self):
    """This method returns the highest grade among all the students if there are any students"""
    if len(self.students) > 0:
      best = None
      for student in self.students:
        if best == None or student.grade > best:
          best = student.grade
      return best
    return False

stud_one = Student("Jack", 78)
stud_two = Student("Jill", 99)
stud_thr = Student("Bob", 94)

class_one = Classroom()
class_one.add_student(stud_one)
class_one.add_student(stud_two)
class_one.add_student(stud_thr)

print(class_one.highest_grade())
````

## Task 11 — Design + object interaction challenge

Create two classes: `Player` and `InventoryManager`.

### `Player`
- constructor parameters:
  - `name`
  - `items`  (this will be a list of strings)

### `InventoryManager`
Add two methods:

1. `add_item(player_obj, item)`
   - append the item to the player's `items` list

2. `trade_first_item(player_a, player_b)`
   - if **both** players have at least one item:
     - swap their first items
     - return `True`
   - otherwise return `False`

### Test
Create two players with different item lists.
Demonstrate both methods.

### Why this is a thinking question
You must manage state stored inside objects, including a mutable list attribute.
Be careful to mutate the correct object data.

````python

# Write your answer here
class Player:
  def __init__(self, name: str, items: list [str]) -> None:
    """This method initializes the attributes of the class"""
    self.name = name
    self.items = items

class InventoryManager:
  def add_item(self, player_obj: Player, item: str):
    """This method adds the given item to the given players item list"""
    player_obj.items.append(item)

  def trade_first_item(self, player_a: Player, player_b: Player):
    """This method swaps two players first items if they both have"""
    if len(player_a.items) > 0 and len(player_b.items) > 0:
      temp = player_a.items[0]
      player_a.items[0] = player_b.items[0]
      player_b.items[0] = temp
      return True
    return False

player_one = Player("First", ["ball"])
player_two = Player("Second", [])
inventory = InventoryManager()
inventory.add_item(player_one, "glove")
inventory.add_item(player_two, "bat")
inventory.trade_first_item(player_one, player_two)
print(player_one.items)
print(player_two.items)
````

## Task 12 — LeetCode-style object design challenge

Create a class named `ScoreTracker`.

### Requirements
- constructor creates an empty dictionary to store player scores
- method `add_score(player_name, points)`
  - if the player is new, create the entry
  - otherwise add to the existing score
- method `top_player()`
  - return the name of the player with the highest score
  - if there are no players, return `None`
- method `top_score()`
  - return the highest score
  - if there are no players, return `None`

### Test
Add several scores for several players and show that repeated calls accumulate correctly.

### Reflection
In a comment, explain why this is a good example of a class:
- what state is being stored?
- what behaviors are being grouped with that state?

````python

# Write your answer here
class ScoreTracker:
  def __init__(self):
    """This method inilializes an empty dictionary attribute for the object"""
    self.scores = {}

  def add_score(self, player_name: str, points: int):
    """This method updates a player and his points."""
    if player_name not in self.scores:
      self.scores[player_name] = points
    else:
      self.scores[player_name] += points

  def top_player(self):
    """This method returns the name of the player with the most points"""
    best = None
    high = None
    for name, score in self.scores.items():
      if high == None or score > high:
        high = score
        best = name
    return best

  def top_score(self):
    """This method returns the highest score across all players"""
    high = None
    for score in self.scores.values():
      if high == None or score > high:
        high = score
    return high

tracker = ScoreTracker()
tracker.add_score("Jack", 10)
tracker.add_score("Bill", 15)
tracker.add_score("Jill", 7)
tracker.add_score("Jack", 21)
tracker.add_score("Bill", 5)
tracker.add_score("Jill", 19)
print(tracker.top_player())
print(tracker.top_score())
````
