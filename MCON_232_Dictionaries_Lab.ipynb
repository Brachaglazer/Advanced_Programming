```python
my_dict = {
            "name" : "Fuzzy",
            "age" : 12,
            "pet-type" : "dog",
            "fur-color" : "brown"
            }

my_dict["eye color"] = "brown"
print(my_dict)
print(f'pet age: {my_dict["age"]}')

print(f'{my_dict["age"]=}')

try:
  print(my_dict["hair"])
except KeyError as ke:
  print("Key not found", ke)
```

```python
# membership test
"age" in my_dict
"hair" not in my_dict
"dog" in my_dict
```

```python
my_dict2 = {}
my_dict2["cat"] = "Polly"
print(my_dict2)
del(my_dict2["cat"])
print(my_dict2)
len(my_dict2)
```

```python
animal = my_dict["pet-type"]
print(animal)
```

```python
# loop on keys
for k in my_dict:
  print(f'{k} value {my_dict[k]}')

# loop on values
for v in my_dict.values():
  print(f'our value is {v}')

# loop on keys and values
for k, v in my_dict.items():
  print(f'keys : {k}, values : {v}')
```

```python
my_dict.get("name")
my_dict.get("last name", "No value there")  # default print if key is missing
```

```python
my_dict.pop("eye color")
my_dict.pop("age")
print(my_dict)
```

```python
import copy

my_dict2 = {"school" : "LAS",
            1601 : "street",
            "num_students" : 100}
print(my_dict2)

my_dict3 = my_dict | my_dict2
print(my_dict3)

my_dict4 = {"college" : "LAS",
            1754 : "street",
            "building" : 100}
my_dict4 |= my_dict3  # +=
print(my_dict4)

my_dict5 = copy.deepcopy(my_dict4)  # deep copy
print(my_dict5)

my_dict6 = my_dict5.copy()  # shallow copy
print(my_dict6)

my_dict6["college"] = my_dict5
print(my_dict6)
print(my_dict6["college"]["college"])
```

```python
list(my_dict.keys())
list(my_dict.values())
list(my_dict.items())

print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())
```

```python
# comprehensions

new = {key : value for key, value in my_dict.items()}
print(new)

nums = {
    1 : "one",
    2 : "two",
    3 : "three",
    4 : "four"
}

even = {key : value for key, value in nums.items() if key % 2 == 0 }
print(even)
```
