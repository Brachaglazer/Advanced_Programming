```python
class Transportation:
  def honk(self):
    return "beep"

class Car(Transportation):
  def __init__(self):
    super().__init__()

  def number_of_passengers(self, num:int):
    return num

  def honk(self):
    return "HONK HONK HONK"

class Truck(Transportation):
  def __init__(self):
    super().__init__()

  def honk(self):
    return "HONK HONK HONK"

my_car = Car()
print(my_car.honk())
print(my_car.number_of_passengers(5))

my_truck = Truck()
print(my_truck.honk())
```
