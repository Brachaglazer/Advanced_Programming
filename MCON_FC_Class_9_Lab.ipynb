```python
class Transportation:
  def honk(self):
    return "beep"

class Car(Transportation):
  def __init__(self):
    super().__init__()

  def number_of_passengers(self, num:int):
    return num

  def honk(self):
    return "HONK HONK HONK"  # overrides parent

class Truck(Transportation):
  def __init__(self):
    super().__init__()

  def honk(self):
    return "CHOO CHOO CHOO"  # return super.honk() will use the Transportation method

class Train(Transportation):
  def __init__(self):
    super().__init__()

# MAIN
def traffic(transportation : Transportation):
  print(transportation.honk())

my_car = Car()
print(my_car.honk())  # checks if child has method and then moves up to check parent
print(my_car.number_of_passengers(5))

my_truck = Truck()
print(my_truck.honk())

my_train = Train()
print(my_train.honk())

traffic(my_car)
traffic(my_truck)
traffic(my_train)
```

ex: method = movement
