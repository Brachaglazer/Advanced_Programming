```python
# If child doesn't add any attributes then parent init is automatic
class Animal:
  def __init__(self, name):
    self.name = name

class Dog(Animal):
  def __init__(self, age, name):
    super().__init__(name)
    self.age = age

my_animal = Dog(12, "Bonzo")
print(my_animal)
```
