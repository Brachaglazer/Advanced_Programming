```python
# If child doesn't add any attributes then parent init is automatic
class Animal:
  def __init__(self, name):
    self.name = name

  def speak(self):
    print("HELLO")

class Dog(Animal):
  def __init__(self, age, name):
    super().__init__(name)
    self.age = age

  def speak(self):
    print("Woof")
    super().speak()

  @staticmethod - # doesn't deal with objects or classes
  # used for its functionality
  def is_age_valid(age:int):
    return 0 < age < 18

my_animal = Dog(12, "Bonzo")
print(my_animal.age, ", ", my_animal.name)
my_animal.speak()
print(Dog.is_age_valid(51))
```

```python
# abstraction = parent defines basic behavior iwthout implementing it and then forces child to implement it
# abstractmethod is method the from parent class
# can think of it as a pleholder
# generating a contract of some time, forcing programmer to program a certain way

# concrete method - already fully written, inherited by children
```

```python
from abc import ABC, abstractmethod, abstractclassmethod # abc is absrtact base class
class Animal(ABC):
  animal_counts = 0

  def __init__(self):
    print("I am forcing you to implement these methods.")
    Animal.animal_counts += 1

  @abstractmethod
  def run(self):
    pass

  @abstractmethod
  def speak(self):
    pass

  @abstractclassmethod
  def animal_count(cls):
    pass

class Dog(Animal):

  def run(self):
    pass # enough to instantialte class

  def animal_count(cls):
    return Dog.animal_counts

  def speak(self):
    pass

dog = Dog();
```

```python
from abc import ABC, abstractmethod

class WordGameController(ABC):
  @abstractmethod
  def play(self):
    pass

  @abstractmethod
  def make_guess(self, letter:str):
    pass

  @abstractmethod
  def check_game_over(self) -> bool:
    pass
```

```python
class TinyWordGame(WordGameController):
  def __init__(self, secret_word:str):
    self.secret_word = secret_word.lower()
    self.guesses = set()

  def play(self):
    print("Welcome to the game, try to guess one letter at a time.")

  def make_guess(self, letter:str):
    letter = letter.lower()
    self.guesses.add(letter)
    return letter in self.secret_word

  def check_game_over(self) -> bool:
    return all(letter in self.guesses for letter in self.secret_word)


game = TinyWordGame("cat")
game.play()
print(game.make_guess("c"))
print(game.check_game_over())
```

```python
class HintWordGame(WordGameController):
  def __init__(self, secret_word:str, hint:str):
    self.secret_word = secret_word.lower()
    self.hint = hint
    self.guesses = set()

  def play(self):
    print("Welcome to the game, here is a hint", self.hint)

  def make_guess(self, letter:str):
    letter = letter.lower()
    self.guesses.add(letter)
    return letter in self.secret_word

  def check_game_over(self) -> bool:
    return all( letter in self.guesses for letter in self.secret_word)

game = HintWordGame("dog", "A floppy companion")
game.play()
print(game.make_guess("c"))
print(game.check_game_over())
```

```python
class LimitWordGame(WordGameController):
  max_attempts = 5
  def __init__(self, secret_word:str):
    self.secret_word = secret_word.lower()
    self.guesses = set()
    self.attempts_used = 0

  @classmethod
  def play(cls):
    print("Welcome to the limited guesses word game! You have", cls.max_attempts, "attempts to guess the word.")

  def make_guess(self, letter: str):
    letter = letter.lower()
    self.guesses.add(letter)
    self.attempts_used += 1
    return letter in self.secret_word

  def check_game_over(self) -> bool:
    return all( letter in self.guesses for letter in self.secret_word) or self.attempts_used >= LimitWordGame.max_attempts

game = LimitWordGame("cat")
game.play()
print(game.make_guess("c"))
print(game.check_game_over())
```

```python
class Shape:
  shape_count = 0

  def __init__(self, color:str) -> None:
    self.color = color
    Shape.shape_count += 1

  @classmethod
  def get_shape_count(cls) -> int:
    return cls.shape_count

  @staticmethod
  def is_valid_color(color:str) -> bool:
    return color in ["red", "blue"]

  @abstractmethod
  def area(self):
    pass

  @abstractmethod
  def describe(self):
    pass
```

```python
class Square(Shape):
  def __init__(self, color:str, length:int) -> None:
    self.length = length
    super().__init__(color)

  def area(self):
    return self.length * self.length

  def describe(self):
    return f"The is a {self.color} square."
```

```python
class Rectangle(Shape):
  def __init__(self, color:str, height:int, width:int) -> None:
    self.height = height
    self.width = width
    super().__init__(color)

  def area(self):
    return self.width * self.height

  def describe(self):
    return f"The is a {self.color} rectangle."
```

```python
square = Square("red", 5)
rectangle = Rectangle("purple", 2, 8)
shapes = [square, rectangle]
for s in shapes:
  print(s.get_shape_count())
  print(s.is_valid_color("red"))
  print(s.area())
  print(s.describe())
```

```python
class Flyer:
  def __init__(self, animal_name:str):
    self.animal_name = animal_name

  def fly(self):
    print("flying through the air")

class Swimmer:
  def __init__(self, swimmer_name:str):
    self.swimmer_name = swimmer_name

  def swim(self):
    print("swimming through the sea")

class FlySwimmer(Flyer, Swimmer):
  def __init__(self, animal_name:str, swimmer_name:str):
    Flyer.__init__(self, animal_name=animal_name)
    Swimmer.__init__(self, swimmer_name=swimmer_name)

duck = FlySwimmer("duck", "duckswimmer")
duck.swim()
duck.fly()
```

```python
class Camera:
  def __init__(self, mode:str):
    self.mode = mode

  def capture(self):
    print(f"taking a picture on {self.mode} mode")

class MusicPlayer:
  def __init__(self, kind:str):
    self.kind = kind

  def play(self):
    print(f"playing {self.kind} music")

class Phone(Camera, MusicPlayer):
  def __init__(self, mode:str, kind:str):
    Camera.__init__(self, mode)
    MusicPlayer.__init__(self, kind)

dumb = Phone("selfie", "pop")
dumb.capture()
dumb.play()
```
