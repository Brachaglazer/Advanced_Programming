## Question 1: Inheritance with Employees

Create a parent class named `Employee`.

### `Employee` must:
- store an employee name
- store a department
- include a method named `work()` that returns a general work message

Then create a child class named `Manager`.

### `Manager` must:
- inherit from `Employee`
- use the parent constructor
- override `work()` so that it returns a manager-specific message

### After writing your classes
Create a `Manager` object and demonstrate that:
- the object stores its data correctly
- the overridden `work()` method is being used


````python
# Write your answer for Question 1 here.
class Employee:
  def __init__(self, name: str, department: str) -> None:
    self.name = name
    self.department = department

  def work(self) -> str:
    return f"{self.name} is assigned to work in the {self.department} department today."

class Manager(Employee):
  def __init__(self, name, department) -> None:
    super.__init__(name, department)

  def work(self):
    return f"{self.name} is the manager of the {self.department} department."
````

````python
manager = Manager("Bill", "Technology")
print(manager.name, manager.department)
print(manager.work())
````


## Question 2: Choose What Belongs in the Parent

Below is a group of possible attributes and behaviors for electronic devices:

- brand
- battery_percent
- charge()
- take_photo()
- make_call()

Create a parent class named `Device` using only the information that should reasonably be shared by many devices.

Then create a child class named `Phone`.

### Requirements
- `Device` should store shared data and include at least one shared behavior
- `Phone` should inherit from `Device`
- `Phone` should add or override behavior that makes sense for a phone

### After writing your classes
Create a `Phone` object and demonstrate that inherited behavior works correctly.


````python
# Write your answer for Question 2 here.
class Device:
  def __init__(self, battery_percent: int) -> None:
    self.battery_percent = battery_percent

    def charge(self):
      self.battery_percent = 100

    def take_photo(self) -> str:
      return f"click"

class Phone(Device):
  def __init__(self, battery_percent: int) -> None:
    super.__init__(battery_percent)

  def make_call(self) -> str:
    return f"Ding a ling"
````

````python
phone = Phone(33)

print(phone.battery_percent)
print(phone.charge())
print(phone.take_photo())
print(phone.make_call())
````


## Question 3: Polymorphism with School Roles

Create a parent class named `SchoolMember` with a method named `introduce()`.

Then create these child classes:
- `Teacher`
- `Student`

Each child class should override `introduce()` in its own way.

### Then:
1. Create one `Teacher` object and one `Student` object
2. Store them together in a list
3. Use one loop to call `introduce()` on each object

## Goal
Your output should clearly demonstrate polymorphism.


````python
# Write your answer for Question 3 here.
class SchoolMember:
  def __init__(self, name: str, floor: int) -> None:
    self.name = name
    self.floor = floor

  def introduce(self) -> str:
    return f"Hi, my name is {self.name}. I spend my time on floor {self.floor}."

class Teacher(SchoolMember):
  def __init__(self, name: str, floor: int) -> None:
    super.__init__(name, floor)

  def introduce(self) -> str:
    return f"Hi, my name is Professor {self.name}. I teach on floor {self.floor}."

class Student(SchoolMember):
  def __init__(self, name: str, floor: int) -> None:
    super.__init__(name, floor)

  def introduce(self) -> None:
    return f"Hi my name is {self.name}. I learn on floor {self.floor}"
````

````python
teacher = Teacher("Sara", 7)
student = Student("Rivka", 3)

school = [teacher, student]

for s in school:
  print(s.introduce())
````


## Bonus (Optional)

Create your own small inheritance example with:
- 1 parent class
- 2 child classes
- 1 overridden method
- 1 loop that demonstrates polymorphism


````python
# Write your bonus answer here.
class Food:
  def __init__(self, name: str, color: str) -> None:
    self.name = name
    self.color = color

  def what_color(self) -> str:
    return f"{self.name} has a perfect {self.color} color."

class Apple(Food):
  def __init__(self, name: str, color: str) -> None:
    super.__init__(name, color)

  def what_color(self) -> None:
    return f"{self.name} comes in a few different colors, very often it is {self.color}."

class Mango(Food):
  def __init__(self, name: str, color: str) -> None:
    super.__init__(name, color)

  def what_color(self) -> None:
    return f"{self.name} is a mix of colors, {self.color} is one of the very noticeable ones."
````

````python
apple = Apple("Golden Crisp Apples", "yellow")
mango = Mango("Mangos", "yellow")

food = [apple, mango]
for f in food:
  print(f.name, f.color)
  print(f.what_color())
````
